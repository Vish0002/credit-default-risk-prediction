# Data prep for Logistic Regression

```python
numeric_features = ["LIMIT_BAL","AGE",'BILL_AMT1', 'BILL_AMT2','BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6', 'PAY_AMT1','PAY_AMT2', 'PAY_AMT3', 'PAY_AMT4', 'PAY_AMT5', 'PAY_AMT6']
categorical_features = ['SEX', 'EDUCATION', 'MARRIAGE']
repayment_features = [ 'PAY_0', 'PAY_2','PAY_3', 'PAY_4', 'PAY_5', 'PAY_6']
```

```python
all_features = numeric_features + categorical_features + repayment_features

print("Total selected:", len(all_features))
print("Unique selected:", len(set(all_features)))
print("Missing columns:", set(X_train.columns) - set(all_features))
print("Unexpected columns:", set(all_features) - set(X_train.columns))
```

Threshold Selection Decision

Selected a provisional threshold of 0.45 for a hypothetical manual credit-risk review system. On the validation set, this achieved 65.60% precision and 36.82% recall, identifying 391 defaults while generating 205 false positives. Compared with threshold 0.30, it requires fewer reviews but misses more defaults. The choice is illustrative because real operational costs and review capacity are unavailable.

# Feature Engineering trail

```python
from sklearn.preprocessing import FunctionTransformer

repayment_cols = ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]
bill_cols = [f"BILL_AMT{i}" for i in range(1, 7)]
payment_cols = [f"PAY_AMT{i}" for i in range(1, 7)]

new_features = [
    "max_delay",
    "late_month_count",
    "recent_late_change",
    "avg_bill_to_limit",
    "avg_payment_amount"
]

def add_credit_features(X):
    X = X.copy()

    if (X["LIMIT_BAL"] <= 0).any():
        raise ValueError("Credit limits must be positive")

    # 1. Highest positive delay code
    X["max_delay"] = X[repayment_cols].clip(lower=0).max(axis=1)

    # 2. Number of months with a documented positive delay
    X["late_month_count"] = (
        X[repayment_cols] >= 1
    ).sum(axis=1)

    # 3. Recent delinquency versus older delinquency
    recent = (X[["PAY_0", "PAY_2", "PAY_3"]] >= 1).sum(axis=1)
    older = (X[["PAY_4", "PAY_5", "PAY_6"]] >= 1).sum(axis=1)

    X["recent_late_change"] = recent - older

    # 4. Average statement amount relative to credit limit
    X["avg_bill_to_limit"] = (
        X[bill_cols].clip(lower=0).mean(axis=1) / X["LIMIT_BAL"]
    )

    # 5. Average payment amount
    X["avg_payment_amount"] = X[payment_cols].mean(axis=1)

    return X
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

engineered_numeric = numeric_features + new_features

fe_preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), engineered_numeric),
        ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_features),
        ("repf", OneHotEncoder(handle_unknown="ignore"), repayment_features)
    ]
)

lgr_fe_model = Pipeline([
    ("features", FunctionTransformer(add_credit_features, validate=False)),
    ("preprocessor", fe_preprocessor),
    ("model", LogisticRegression(max_iter=1000))
])
```

```python
# Cross-validation: training data only
fe_cv_scores = cross_val_score(
    lgr_fe_model,
    X_train,
    y_train,
    cv=cv,
    scoring="average_precision"
)

lgr_fe_model.fit(X_train, y_train)

fe_prob = lgr_fe_model.predict_proba(X_val)[:, 1]

fe_ap = average_precision_score(y_val, fe_prob)
fe_roc = roc_auc_score(y_val, fe_prob)

print("CV mean AP:", fe_cv_scores.mean())
print("CV AP std:", fe_cv_scores.std())
print("Validation AP:", fe_ap)
print("Validation ROC-AUC:", fe_roc)
```
